```python
num1 = float(input("Digite seu número: "))
num2 = float(input("Digite seu outro nímero: "))
soma = num1 + num2
print("O resultado será:", soma)
```

```python
#  Atividade 2 - Subtração
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))
Sub = n1 - n2
print("A subtração de n1 com n2 é:", Sub)
```

```python
#  Atividade 3 - Operações completas
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))
Soma = n1 + n2
Subtracao = n1 - n2
Divisao = n1/n2
Multiplicacao = n1*n2
print("Soma:", Soma, "\nSubtração:", Subtracao, "\nMultiplicação:", Multiplicacao, "\nDivisão:", Divisao)
```
