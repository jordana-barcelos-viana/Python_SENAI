```python
idade = int(input("Digite sua idade: "))
if idade >= 18:
  print("Apto para voto")
else:
  print("Não apto")
```

```python
# 1 - Criar um sistema de senha. A senha deve ser senai2026. Se for esta senha, aparece "LIBERADO". Se não for, "NÃO LIBERADO".
# 2 - Meta: Cada vendedor deve ter meta de R$8000,00. Se aparecer, bateu a meta. Se não, não atingiu a meta.
```

```python
# 1 -

senha = input("Digite a senha: ")
if senha == "senai2026":
  print("LIBERADO")
else:
  print("ACESSO NÃO LIBERADO")
```

```python
# 2 -

meta = float(input("Faturamento do vendedor: (Digite apenas números)"))
if meta >= 8000:
  print("Bateu a meta")
else:
  print("Não atingiu a meta")
```
